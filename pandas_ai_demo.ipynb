# Vast CSV Analyst demo

This notebook only prepares and launches the demo. **Dataset selection, Fast/Deep mode, questions, follow-ups, tables, and charts all happen in the Streamlit UI.**

Before starting, use Jupyter's file browser to make sure these are in `/workspace`:

- `app.py`
- One or more CSV files

The Vast template should already be serving Qwen through vLLM on port `18000`.

## 1. Install the demo environment

Run this cell once on a new Vast instance. PandasAI needs Python 3.11, so the cell creates a small environment, installs the demo packages, and adds a Jupyter kernel named **Python 3.11 (PandasAI)**.

```python
from pathlib import Path
import subprocess

venv = Path("/workspace/pandasai-venv")
python = venv / "bin/python"

if not python.exists():
    subprocess.run(["uv", "venv", "--python", "3.11", str(venv)], check=True)

subprocess.run([
    "uv", "pip", "install", "--python", str(python),
    "ipykernel", "pandasai", "pandasai-litellm", "streamlit",
], check=True)
subprocess.run([
    str(python), "-m", "ipykernel", "install", "--user",
    "--name", "pandasai-311", "--display-name", "Python 3.11 (PandasAI)",
], check=True)
print("Installed Python 3.11 (PandasAI). Follow step 2 below.")
```

## 2. Refresh Jupyter and switch kernels

After the install finishes:

1. **Refresh this Jupyter page.**
2. Choose **Kernel → Change Kernel → Python 3.11 (PandasAI)**.
3. Run the short confirmation cell below.

Jupyter will remember the selected kernel after the notebook is saved.

```python
import sys

print(f"Using Python {sys.version.split()[0]}")
print("Ready to launch the UI." if sys.version_info[:2] == (3, 11) else "Switch to Python 3.11 (PandasAI) before continuing.")
```

## 3. Launch Streamlit and open the tunnel

Run the final cell and **leave it running** while you use the demo. Streamlit will print that it is available on port `8501`.

Then open the **Vast Instance Portal**:

1. Open **Tunnels**.
2. Create a tunnel to `http://localhost:8501`.
3. Open the tunnel link Vast gives you.
4. Choose a CSV and Fast or Deep mode in the sidebar, then start chatting. Ask follow-up questions naturally; Deep mode is recommended for charts or more involved analysis.

Charts generated by the demo are saved in `/workspace/charts`. To stop Streamlit, interrupt the final notebook cell.

```python
import subprocess
import sys

subprocess.run([
    sys.executable, "-m", "streamlit", "run", "/workspace/app.py",
    "--server.address", "127.0.0.1",
    "--server.port", "8501",
    "--server.headless", "true",
    "--browser.gatherUsageStats", "false",
], check=True)
```
